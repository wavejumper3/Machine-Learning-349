# Contrast how hard voting aggregates predictions versus soft voting. Why do well-calibrated probability estimates matter when implementing soft voting across heterogeneous models?

Hard voting votes by having each classifier output a hard categorical value, and whichever value is output the msot wins. Soft voting has each model output a probability value across a continuus range, or a group of probability values, that are averaged out. The highest average wins, and is selected. Soft voting relies on all the models outputting the same probability ranges, which means that models outputting wildly different maximum and minimum ratings effectiely weights the models implicitly, which defeats much of the purpose of using heterogenous model groupings for voting. # What is the role of out-of-bag evaluation in Bagging classifiers, and why does an OOB score serve as an efficient alternative to traditional cross-validation splits during ensemble training?

Each individual model has its own train/test split. The "out of bag" values for any given model are in its test split. By using a large amount of models, any given data point will be in the training group for many of them, and the test group for some. The out-of-bag evaluation means checking each data point against all the trees that did NOT train on that data point. The efficency of this is that it works specifically as a test for new data, like tradition cross-validation splits, while not requiring training more models than needed, unlike a traditional cross-validation split. A traditional cross validator creates a group of models trained on different subsets of the data, but only uses one. Bagging uses all of them anyways, making sure all of that compute is put to good use.

# How does introducing random feature subsetting in Random Forests differ from standard Bagging, and how does this feature diversity impact model variance and generalization?

In standard bagging, each tree has access to all of the input features. This means that overall, each tree will likely choose the same input features as most valuable, which reduces model variance. Subsetting means each tree only has access to a subset of the input features. This means that the "most correlated input features" won't be present in all trees, forcing trees to use the values they otherwise would be unlikely to pick. This leads to a wider variety of trees, and a more generalized total Bagging model.
